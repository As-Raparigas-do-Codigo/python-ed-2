*   Sequência de instruções que pode ser chamada infinitas vezes
*   Existem funções já definidas no python, exemplo: print() ou o int()

*   Para definir funções em python usamos a keyword *def* e terminamos com parenteses e dois pontos
*   Todo o conteúdo da função, depois dos dois pontos, tem que estar identado

**Return Values**

As funções também podem retornar valores, qualquer tipo de variável existente no python pode ser devolvido por uma função. Usamos a keyword *return*

```python
from builtins import print
def somar_1_2():
  return 1 + 2

print(somar_1_2())
```

```python
from builtins import print
def dizer_ola_raparigas():
  return "Olá Raparigas do Código!"

print(dizer_ola_raparigas())
```

```python
from builtins import print
def maior_que_10():
  a = 15
  return a > 10

print(maior_que_10())
```

# Módulos

Há funções que já existem no Python e não temos que fazer nada explicitamente no nosso código para as usar.
Por exemplo: *print*, *max*, *min*, *len*.

```python
print("Esta é a função print")
print("A função max devolve o máximo entre um conjunto de variáveis do mesmo tipo: %d" % max(1, 10, 3, 51))
print("A função max devolve o mínimo entre um conjunto de variáveis do mesmo tipo: %d" % min(10, 20, 40, 1))
print("A função len devolve o tamanho de uma lista: %d" % len([10, 20, 20]))




print(max("10", "20", "40", "1000", "-1", "-3.0", "6.25", "9000000", "ZXV", "ZZ"))
print(min(10, 20, 40, 1000, 1, 3.0, 6.25, 9000000, -4.000))
print(len([30, 50, 100, 1, 1, 1, 1, 1, 1, 1, 1]))

print("----")
a = 10
print("A " + str(a) + " B " + str(a))
nome = "Ana"
print("A %s gosta do numero %d" % (nome, a))

l = [30, 50, 100, 1, 1, 1, 1, 1, 1, 1, 1]
media = sum(l)/len(l)
print(media)

```

Será que não existem outros comportamentos, ou melhor, outras funções, já definidas no Python?

A resposta é sim! 
O Python é uma linguagem que tem muitas funções, definições e instruções já definidas que podemos aproveitar quando estamos a desenvolver o nosso código. Muitas dessas já vêm com a instalação do Python, como as que vamos usar, e outras mais específicas temos que instalar antes de usar.

Em Python, esses conjuntos de funções chamam-se **módulos**.

Então, como ir buscar esses módulos?

```python
import random

rnd = random.Random() # Variável para gerar números aleatórios
numero_aleatorio = rnd.random() # Número aleatório entre 0 e 1
print(numero_aleatorio)

dado = rnd.randrange(1, 7) # Número inteiro aleatório entre 1 e 6
print(dado)

aleatorio_impar = rnd.randrange(1, 100, 2)
print(aleatorio_impar)
```

**Módulo time**

```python
import time

def somar_ate(numero):
  a = 0
  for i in range(numero):
    a += i
  
  return a

inicio = time.perf_counter()
soma = somar_ate(100000000)
fim = time.perf_counter()

print("soma = %d" % soma)
print("tempo total: %.5f - %.5f = %.5f segundos" % (fim, inicio, fim - inicio))
```

Como saber qual a unidade de tempo que a função time.perf_counter() retorna?

Temos que ler a documentação da função. Podemos ver no IDE que estivermos a usar, se tiver essa informação, ou pesquisar o nome da função no Google.

```python
from time import perf_counter, perf_counter_ns

def somar_ate_v2(numero):
  a = 0
  for i in range(numero):
    a += i
  
  return a

inicio_2 = perf_counter()
soma_2 = somar_ate_v2(100000000000)
fim_2 = perf_counter()

tempo_ns = perf_counter_ns()


print("soma = %d" % soma_2)
print("tempo total: %.5f - %.5f = %.5f" % (fim_2, inicio_2, fim_2 - inicio_2))
```

# Módulo *math*

O módulo *math* tem muitas operações matemáticas úteis já programadas.

```python
import math

# Constantes já definidas no módulo math
pi = math.pi # o pi
e = math.e # o número de euler/nepper

print("pi = %.5f\te = %.5f" % (pi, e))
```

```python
import math
# Funções matemáticas já definidas no módulo math
n = 4
raiz_quadrada = math.sqrt(n)
print("sqrt(%.1f) = %.1f" % (n, raiz_quadrada))

ang_rads = pi
ang_graus = math.degrees(ang_rads)
print("%.3f Rad = %.1f graus" % (pi, ang_graus))

seno = math.sin(math.radians(ang_graus))
print("seno(%.1f) = %.2f" % (ang_graus, seno))

coseno = 0
ang_rads_2 = math.acos(coseno)
print("acos(%.2f) = %.3f Rad = %.1f graus" % (coseno, ang_rads_2, math.degrees(ang_rads_2)))
```
