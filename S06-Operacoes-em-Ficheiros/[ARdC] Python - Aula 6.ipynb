# Operações em ficheiros de texto: escrita

Para guardar dados num ficheiro, usamos a operação de escrita. Em Python, usa-se a função *open*:

```
my_file = open("teste.txt", "w")
```

* O primeiro argumento da função open é uma *string* com o nome do ficheiro (neste caso, "teste.txt") e o segundo argumento é o tipo de operação, neste caso, escrita ("w", write);

* A função open retorna um tipo de variável ficheiro, com propriedades específicas que permite fazer operações sobre o ficheiro de texto "teste.txt";

* Ao usar o modo escrita "w", o ficheiro é substituído se já existir, ou criado e depois aberto se ainda não existir.



````python
ficheiro = open("teste.txt", "w")
````

Ao abrir o explorador de ficheiros do Notebook, é possível ver que, depois de correr a linha acima, é criado um ficheiro vazio com o nome "teste.txt".

E agora, como escrever conteúdo no ficheiro?

````python
ficheiro.write("Isto é um ficheiro!\n")
ficheiro.write("-------------------\n")
ficheiro.write(".-.-.-.-.-.-.-.-.-.\n")
````

No final das operações, deve sempre fechar-se o acesso ao ficheiro.

````python
ficheiro.close()
````

**Num ficheiro só é possível escrever strings.**

# Operações em ficheiros de texto: leitura

Para ler um ficheiro, usa-se a opção "r" (read).


````python
with open("numeros.txt", "r") as ficheiro_numeros:
  for linha in ficheiro_numeros:
    print(linha)
````

**with**

Neste exemplo usa-se a keyword *with* para definir a variável usada para as operações no ficheiro.

Ao usar este formato, não é necessário fechar o acesso ao ficheiro, já que isso é feito automaticamente.



**for**

Ao aplicar o for a um ficheiro, itera-se por cada linha do ficheiro.

````python
mais_um_ficheiro = open("file_file", "r")
````

Não é possível abrir em modo leitura um ficheiro que não existe.

## Ficheiro como lista

É possível guardar todo o conteúdo de um ficheiro numa lista, em que cada item é uma linha do **ficheiro**.

````python
with open("numeros.txt", "w") as ficheiro_temp:
  for i in range(100):
    ficheiro_temp.write(str(i) + "\n")
````

````python
with open("numeros.txt", "r") as ficheiro_numeros:
  todas_linhas = ficheiro_numeros.readlines()

print(todas_linhas)
````

Agora, a variável *todas_linhas* é uma lista com cada linha do ficheiro "numeros.txt" (sempre do tipo string).

## Ficheiro como string

Também é possível guardar todo o ficheiro numa única string, com a função *read*.

````python
with open("numeros.txt", "r") as ficheiro_numeros:
  ficheiro_completo = ficheiro_numeros.read()

print(ficheiro_completo)
````

# Outras operações em ficheiros

Além do modo escrita ("w") e do modo leitura ("r"), existem outros modos de acesso a ficheiros:

* "r": abrir para leitura (dá erro se o ficheiro não existir);

* "w": criar um novo ficheiro (**substitui** o conteúdo se já existir);

* "a": criar ou abrir um ficheiro (**acrescenta** o conteúdo se já existir);

* "x": criar um novo ficheiro (dá erro se o ficheiro já existir).

# Armazenamento de dados em formato CSV

## O que é o formato CSV?

* *Comma-Separated Values*;
* Permite organizar uma grande quantidade de dados num ficheiro de texto, simples de ler e escrever;
* Ficheiro de texto com uma lista de dados, tipicamente separados por vírgula em cada linha;
* Os ficheiros CSV têm a extensão .csv, mas são meros ficheiros de texto.


Exemplo:


```
nome, idade, email
andre, 25, andre@raparigasdocodigo.pt
beatriz, 34, beatriz@raparigasdocodigo.pt
catarina, 28, catarina@raparigasdocodigo.pt
...
```



## Leitura e escrita do formato CSV: Lista

O módulo de Python *csv* permite efectuar operações sobre este tipo de ficheiro, como escrita linha por linha.


````python
import csv

with open("dados.csv", "w") as ficheiro_csv:
  csv_escrita_lista = csv.writer(ficheiro_csv, delimiter=',', quotechar='"', quoting=csv.QUOTE_MINIMAL)

  csv_escrita_lista.writerow(["cidade", "latitude", "longitude", "populacao"])
  csv_escrita_lista.writerow(["Lisboa", 38.7223, 9.1393, 504718])
  csv_escrita_lista.writerow(["Porto", 41.1579, 8.6291, 214349])
  csv_escrita_lista.writerow(["Coimbra", 40.2033, 8.4103, 105842])
  csv_escrita_lista.writerow(["Leiria", 39.7495, 8.8077, 126879])
  csv_escrita_lista.writerow(["Évora", 38.5714, 7.9135, 56596])

````

A função *reader* do módulo *csv* permite a leitura do ficheiro, e depois aceder a uma lista (a variável csv_linha abaixo) com todos os valores de cada linha.

````python
import csv

with open("dados.csv") as ficheiro_csv:
    csv_reader_list = csv.reader(ficheiro_csv, delimiter=",")
    nr_linha = 0
    for csv_linha in csv_reader_list:
        if nr_linha > 0:
            print("A cidade %s tem %s habitantes" % (csv_linha[0], csv_linha[3]))
        nr_linha += 1
    
    print("\nO ficheiro tem %d linhas de dados." % (nr_linha-1))
````

## Leitura e escrita do formato CSV: Dicionário

No exemplo anterior, o conteúdo do ficheiro CSV foi escrito de e lido para uma variável do tipo lista. 

O módulo *csv* também permite ler e escrever para uma variável do tipo dicionário, em que a chave para cada valor é o nome da coluna.

````python
import csv

with open("dados2.csv", mode="w") as ficheiro_csv:
    colunas = ["cidade", "latitude", "longitude", "populacao"]
    csv_escrita_dict = csv.DictWriter(ficheiro_csv, fieldnames=colunas, delimiter=',', quotechar='"', quoting=csv.QUOTE_MINIMAL)

    csv_escrita_dict.writeheader()
    csv_escrita_dict.writerow({"cidade": "Lisboa", "latitude": 38.7223, "longitude": 9.1393, "populacao": 504718})
    csv_escrita_dict.writerow({"cidade": "Porto", "latitude": 41.1579, "longitude": 8.6291, "populacao": 214349})
    csv_escrita_dict.writerow({"cidade": "Coimbra", "latitude": 40.2033, "longitude": 8.4103, "populacao": 105842})
    csv_escrita_dict.writerow({"cidade": "Leiria", "latitude": 39.7495, "longitude": 8.8077, "populacao": 126879})
    csv_escrita_dict.writerow({"cidade": "Évora", "latitude": 38.5714, "longitude": 7.9135, "populacao": 56596})
````

````python
import csv

with open("dados2.csv") as ficheiro_csv:
    csv_reader_dict = csv.DictReader(ficheiro_csv, delimiter=",")
    nr_linha = 0
    for csv_dict in csv_reader_dict:
        print("A cidade %s tem %s habitantes" % (csv_dict["cidade"], csv_dict["populacao"]))
        nr_linha += 1
    
    print("\nO ficheiro tem %d linhas de dados." % nr_linha)
````

# Exceções



## O que são exceções?

Um erro na execução de um programa gera uma exceção.

**Exceções não são erros de sintaxe.**

Exemplos de erros que geram exceções:

*   Dividir um número por zero;
*   Aceder a um item inexistente numa lista;
*   Tentar ler um ficheiro que não existe;
*   Entre muitos outros...

````python
a = 20/0
````

````python
lista = [1, 2, 3]
lista[10]
````

````python
ficheiro_falso = open("ficheiro_falso.txt", "r")
````

Para lidar com estes erros, pode usar-se a estrutura de Python *try*.

A forma completa do *try* é a seguinte, em que o *except*, o *else* e o *finally* são opcionais.
```
try:
  # Codigo
except Error:
  # Codigo
else:
  # Codigo
finally:
  # Codigo
```



````python
try:
  a = 20/0
except ZeroDivisionError:
  print("Divisão por zero não é permitida")
else:
  print("Não ocorreu nenhuma exceção")
finally:
  print("Este código corre sempre")
````

````python
try:
  lista = [1, 2, 3]
  lista[10]
except ZeroDivisionError:
  print("Não é possível aceder a este índice")
else:
  print("Não ocorreu nenhuma exceção")
finally:
  print("Este código corre sempre")
````

````python
try:
  ficheiro_falso = open("ficheiro_falso.txt", "r")
except IndexError:
  print("Não é possível aceder a este índice")
except ZeroDivisionError:
  print("Divisão por zero não é permitida")
except FileNotFoundError:
  print("O ficheiro não existe")
else:
  print("Não ocorreu nenhuma exceção")
finally:
  print("Este código corre sempre")
````

## Como gerar exceções?

É possível gerar exceções em qualquer ponto do código, por exemplo numa função.

Ao ser lançada a exceção, a execução do código pára. No exemplo abaixo, a linha `print("Exceção")` nunca é corrida porque aparece sempre depois da exceção.

Existem muitas exceções já definidas no Python, por exemplo:

* ValueError;
* ZeroDivisionError;
* FileNotFoundError;
* ImportError;
* IndexError;
* KeyError;
* KeyboardInterrupt;
* Entre muitas outras...

````python
def imprimir_idade(idade):
  if type(idade) is not int or idade < 0:
    raise ValueError("O valor inserido não é válido!")
    print("Exceção")
  else:
    print("Idade válida: %d" % idade)
````

````python
imprimir_idade("A")
````

````python
imprimir_idade(10)
````

````python
imprimir_idade(-3)
````

````python
try:
  imprimir_idade([1, 2, 3])
except ValueError as e:
  print("Erro: %s" % e)
````
