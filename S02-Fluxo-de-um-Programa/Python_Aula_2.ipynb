### **Relembrando os Operadores de Comparação**
O resultado da operação de comparação é um valor booleano: **True** caso a comparação seja verdadeira, **False** caso seja falsa.
* Igual: **==**
* Diferente: **!=**
* Maior: **>**
* Menor: **<**
* Maior ou igual: **>=**
* Menor ou igual: **<=**

````python
x = 7.4
print("x == 7.4: " + str(x == 7.4))
print("x == 5.4: " + str(x == 5.4))
print("x != 5.4: " + str(x != 5.4))
print("x != 7.4: " + str(x != 7.4))
print("x > 10.4: " + str(x > 10.4))
print("x > 5.4: " + str(x > 5.4))
print("x < 5.4: " + str(x < 5.4))
print("x < 10.4: " + str(x < 10.4))
print("x >= 7.4: " + str(x >= 7.4))
print("x <= 4.4: " + str(x <= 4.4))
````

---
**Relembrando Operadores Lógicos**<br>
Os operadores lógicos permitem combinar várias expressões booleanas (e.g., resultantes de operações de comparação).
* Todas as expressões booleanas têm de ser verdadeiras: **and**
* Apenas uma das expressões booleanas tem de ser verdadeira: **or**
* Inverte o resultado booleano: **not**



````python
x = 7.4
print("x > 7 and x < 10: " + str(x > 7 and x < 10))
print("x > 4.4 and x < 6.4: " + str(x > 4.4 and x < 6.4))
print("x > 7 or x < 10: " + str(x > 7 or x < 10))
print("x > 4.4 or x < 6.4: " + str(x > 4.4 or x < 6.4))
print("x > 10.4 or x < 3.4: " + str(x > 10.4 or x < 3.4))
print("not(x > 7): " + str(not(x > 7)))
print("(not(x > 7) and x < 7) or x > 4: " + str((not(x > 7) and x < 7) or x > 4))
````

### **Execução Condicional: if**
Uma instrução condicional permite correr um bloco de código apenas se uma condição for verdadeira.
A forma mais simples de escrever uma instrução condicional é com a instrução if.

A palavra if é seguida de uma condição que quando for verdadeira irá executar o bloco de código que a segue:

```
if CONDIÇÃO:
    BLOCO DE CÓDIGO
```


````python
x = 6
if x % 2 == 0:
  print(str(x) + " é par")
````

### **Bloco de Código e Indentação**
Um bloco de código é uma ou mais instruções que devem ser executadas uma após a outra, de cima para baixo da esquerda para a direita, e que estão ao mesmo nível de indentação.

````python
print("Soma de 2 com 5 =", str(2 + 5))
print("Subtração de 10 com 5 =", str(10 - 5))
````

A indentação de uma linha é o seu avanço em relação à margem. Normalmente é usado uma tab, ou 4 espaços em branco para aumentar a identação.

No caso do nosso if anterior, a linha 

```
    print(str(x) + " é par")
```
encontra-se indentada.
O que acontece se removermos a indentação?

````python
x = 7
if x % 2 == 0:
print(str(x) + " é par")
````

Ocorre um erro de indentação (IdentationError).
Olhando com mais detalhe para o nosso erro, podemover que que após a intrução if é esperado um bloco indentado.

### **Condição Alternativa: else**

Um segundo formato da instrução if é a execução alternativa, na qual existem duas possibilidades e a condição determina qual delas será executada.


````python
x = 7
if x % 2 == 0:
    print(str(x) + " é par")
else:
    print(str(x) + " é ímpar")
````

image.png



**Condicionais Encadeados: elif**

Às vezes existem mais de duas possibilidades e precisamos de mais que dois (if-else) ramos. Para esses casos podemos criar um novo ramo elif. Elif é uma abreviação de else if que significa "senão se".

````python
x = -1

eNegativo = x < 0
ePar = x % 2 == 0

if eNegativo:
    print(str(x) + " é um número negativo")
elif ePar:
    print(str(x) + " é par")
else:
    print(str(x) + " é ímpar")
````

image.png


**Condicionais Aninhados**

O condicional pode ser aninhado dentro de outro, sendo que este só vai ser avaliado se o de fora fora

````python
x = 4
if x > 0 :
    if x < 10:
        print(str(x) + " é positivo e de apenas um algarismo")
    else:
        print(str(x) + " é positivo e de múltiplos um algarismo")
````

image.png

**Simplificação de Condicionais Aninhados usando Operadores Lógicos**

Os Operadores Lógicos como o and e o or fornecem uma maneira de simplificar instruções condicionais aninhadas. Por exemplo, podemos reescrever o código anterior numa única condicional:

````python
x = 4
if x > 0 and x < 10:
    print(str(x) + " é positivo e de apenas um algarismo")
elif x >= 10:
    print(str(x) + " é positivo e de múltiplos um algarismo")
````

````python
x = 4
if 0 < x < 10:
    print(str(x) + " é positivo e de apenas um algarismo")
elif x >= 10:
    print(str(x) + " é positivo e de múltiplos um algarismo")
````
