### **Variáveis e Tipos de Dados Simples**

Podemos aceder ao resultado devolvido pelo input() se o guardarmos numa variável.<br>
As variáveis servem para armazenar em memória os dados de um programa.



````python
idade = input("Qual a sua idade? ")
print("O utilizador tem " + idade + " anos.")
````

Os nomes das variáveis obedecem a um conjunto de regras:
*   Só podem conter letras, números e *underscores*.
*   Devem começar por uma letra ou por *underscore*.
*   São *case-sensitive*, ou seja, “idade” é diferente de “IDADE”.
*   Por convenção, quando contêm várias palavras estas são separadas por *underscore*.
*   Não podem ser iguais a palavras reservadas do Python.



````python
idade = input("Qual a sua idade? ")
IDADE = input("Qual a sua idade? ")
print("As idades são " + idade + " e " + IDADE + " anos.")
````

---
Será que as variáveis só podem armazenar texto? Não!<br>
As variáveis podem armazenar diversos tipos de dados:
* Texto: **str**
* Números Inteiros: **int**
* Números Reais: **float**
* Números Complexos: **complex** 
* Booleano: **bool**
* Entre outros que iremos estudar no futuro...

O Python infere o tipo de dados das variáveis através do valor atribuído.<br>
Nós podemos saber o tipo de dados de uma variável através da função **type()**.



````python
varTexto = "Isto é uma frase..."
varInt = 7
varFloat = 3.14
varComplex = 1+2j
varBool = True

print(type(varTexto))
print(type(varInt))
print(type(varFloat))
print(type(varComplex))
print(type(varBool))
````

---
Por vezes temos necessidade de converter o tipo de dados de uma variável (**type casting**).<br>
Por exemplo, vamos pedir ao utilizador o raio de um círculo e calcular a sua área.


````python
raio = input("Raio do círculo: ")
area = 3.14 * raio * raio
print("Área do círculo: " + area)
````

O bloco anterior deu um erro na linha do cálculo porque o raio está em formato de texto (**str**).<br>
Para que o cálculo seja possível temos de converter esse raio para um número real (**float**).


````python
raio = input("Raio do círculo: ")
raio = float(raio)
area = 3.14 * raio * raio
print("Área do círculo: " + area)
````

Temos agora um erro na função print(), que nos diz que não é possível concatenar texto com um número real.<br>
O valor da área, para ser concatenado com o restante texto, tem de ser convertido para **str**.


````python
raio = input("Raio do círculo: ")
raio = float(raio)
area = 3.14 * raio * raio
print("Área do círculo: " + str(area))
````

---
Ao nível das strings (i.e., texto) podemos fazer diversas operações de formatação/modificação.<br><br>
**Dica:** Podemos acrescentar comentários ao nosso código com o símbolo #.<br>Os comentários aumentam a legibilidade do código e facilitam a sua interpretação.


````python
texto = "Olá, mundo!"
print(texto[:3]) # Obter os caracteres entre a 1ª e a 3ª posições
print(texto[5:]) # Obter os caracteres a partir da 6ª posição
print(texto[2:6]) # Obter os caracteres entre a 3ª e a 6ª posições
print(texto[1]) # Obter o caracter da 2ª posição 
print(texto[-1]) # Obter o último caracter
print(texto.upper()) # Converter todos os caracteres para maiúsculas 
print(texto.lower()) # Converter todos os caracteres para minúsculas 
print(texto.replace("!", "?")) # Substituir o '!' por '?'
print(len(texto)) # Tamanho da string
print("PI = {:.2f}".format(3.14159)) # Arredondar um número real a 2 casas decimais
````

### **Expressões e Operadores**

Uma expressão é um conjunto de valores, variáveis, operadores e/ou invocações a funções, cuja combinação irá produzir um resultado.<br>
Se pensarmos na soma de uma variável com um valor, temos uma expressão cujo operador é a soma. Existem diversos operadores com prepósitos diferentes: aritméticos, de atribuição, de comparação, lógicos, e outros que iremos estudar no futuro.

---

**Operadores Aritméticos**
* Soma: **+**
* Subtração: **-**
* Multiplicação: **\***
* Divisão: **/**
* Módulo: **%**
* Exponenciação: **\*\***
* Divisão Inteira: **//**

````python
x = 7.4
print("Soma:", end=" ")
print(x + 3) # Soma
print("Subtração:", end=" ")
print(x - 3) # Subtração
print("Multiplicação:", end=" ")
print(x * 3) # Multiplicação
print("Divisão:", end=" ")
print(x / 3) # Divisão
print("Módulo:", end=" ")
print(x % 3) # Módulo
print("Exponenciação:", end=" ")
print(x ** 3) # Exponenciação
print("Divisão Inteira:", end=" ")
print(x // 3) # Divisão Inteira
````

---
**Operadores de Atribuição**

Para além do operador de atribuição *standard* ( **=** ), podemos aplicar operações aritméticas no processo de atribuição.<br>
Por exemplo, **x += 3** é igual a termos **x = x + 3**. O mesmo se aplica aos restantes operadores aritméticos.


````python
x = 7.4
x += 3
print(x)
x *= 3
print(x)
````

---
**Operadores de Comparação**<br>
O resultado da operação de comparação é um valor booleano: **True** caso a comparação seja verdadeira, **False** caso seja falsa.
* Igual: **==**
* Diferente: **!=**
* Maior: **>**
* Menor: **<**
* Maior ou igual: **>=**
* Menor ou igual: **<=**

````python
x = 7.4
print("x == 7.4: " + str(x == 7.4))
print("x == 5.4: " + str(x == 5.4))
print("x != 5.4: " + str(x != 5.4))
print("x != 7.4: " + str(x != 7.4))
print("x > 10.4: " + str(x > 10.4))
print("x > 5.4: " + str(x > 5.4))
print("x < 5.4: " + str(x < 5.4))
print("x < 10.4: " + str(x < 10.4))
print("x >= 7.4: " + str(x >= 7.4))
print("x <= 4.4: " + str(x <= 4.4))
````

---
**Operadores Lógicos**<br>
Os operadores lógicos permitem combinar várias expressões booleanas (e.g., resultantes de operações de comparação).
* Todas as expressões booleanas têm de ser verdadeiras: **and**
* Apenas uma das expressões booleanas tem de ser verdadeira: **or**
* Inverte o resultado booleano: **not**



````python
x = 7.4
print("x > 7 and x < 10: " + str(x > 7 and x < 10))
print("x > 4.4 and x < 6.4: " + str(x > 4.4 and x < 6.4))
print("x > 7 or x < 10: " + str(x > 7 or x < 10))
print("x > 4.4 or x < 6.4: " + str(x > 4.4 or x < 6.4))
print("x > 10.4 or x < 3.4: " + str(x > 10.4 or x < 3.4))
print("not(x > 7): " + str(not(x > 7)))
````

---
**Ordem das Operações**<br>

Quando uma expressão tem mais do que um operador, a ordem de aplicação rege-se pelas **regras de precedência**:
1.   Parênteses têm a máxima precedência
2.   Exponenciação
3.   Multiplicação e divisão
4.   Adição e subtração
5.   Operadores com a mesma precedência são aplicados da esquerda para a direita (exceto a exponenciação)


````python
print("2 * (3 - 1): " + str(2 * (3 - 1)))
print("(1 + 1) ** (5 - 2): " + str((1 + 1) ** (5 - 2)))
print("2 ** 1 + 1: " + str(2 ** 1 + 1))
print("3 * 1 ** 3: " + str(3 * 1 ** 3))
print("2 * 3 - 1: " + str(2 * 3 - 1))
print("5 - 2 * 2: " + str(5 - 2 * 2))
print("6 - 3 + 2: " + str(6 - 3 + 2))
print("2 ** 3 ** 2: " + str(2 ** 3 ** 2))
print("(2 ** 3) ** 2: " + str((2 ** 3) ** 2))
````
