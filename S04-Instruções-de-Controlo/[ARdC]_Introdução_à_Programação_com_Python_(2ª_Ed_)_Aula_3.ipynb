###**Iteração**
*   Quando repetimos um conjunto de instruções várias vezes chama-se de iteração.
*   A iteração pode ajudar-nos a simplificar bastante o nosso código

Vamos ver o código seguinte para imprimir a tabuada do 2:




```python
print("2 x 1 =", 2*1)
print("2 x 2 =", 2*2)
print("2 x 3 =", 2*3)
print("2 x 4 =", 2*4)
print("2 x 5 =", 2*5)
print("2 x 6 =", 2*6)
print("2 x 7 =", 2*7)
print("2 x 8 =", 2*8)
print("2 x 9 =", 2*9)
print("2 x 10 =", 2*10)
```

* Para iterarmos a nossa tabuada, temos de definir uma variável inicial que começa em 1 e vai iterar, incrementando sempre em 1
* Vejamos como podemos alterar os valores que tínhamos definido para uma variável inicialmente.

```python
a = 1
b = a 
print("A é igual a B?",a==b)
a = 2  
print("A é igual a B?",a==b)
```

*   O novo valor da variável pode depender do valor anteriormente definido.



```python
a = 1
print("2 x",a,"=", 2*a)
a = a+1
print("2 x",a,"=", 2*a)
```

Já sabemos como preparar o nosso código para a iteração, agora falta-nos aprender os tipos de ciclo para indicarmos ao computador quando deve iterar!

###**Ciclo While**
* Uma estrutura de repetição permite executar um bloco de instruções várias vezes enquanto determinada condição é verdadeira.

```python
tabuada = 2
resultado = 0
iteracao = 1

while (iteracao <= 10):
  resultado += tabuada
  print("2 x",iteracao,"=", resultado)
  input()
  iteracao += 1
```

```python
iteracao = 1
while(iteracao <= 10):
    print("2 x",iteracao,"=", 2*iteracao)
    iteracao += 1
```

###**Ciclos Infinitos**
* Se a condição que estamos a verificar nunca for falsa, o ciclo pode acontecer um número infinito de vezes, nunca acaba!
* As instruções do bloco de instruções têm que garantir que o progresso do ciclo é tal que a determinada altura a condição se torna falsa.

```python
n = 2
while (n % 2 == 0):
  n += 2
```

```python
i = 0
while (True):
  i += 1
  print(i)
```

###**Ciclo For**
Uma estrutura de repetição que permite processar todos os elementos de uma lista. Em cada iteração, o próximo elemento da lista é atribuido à variável do ciclo antes do corpo do ciclo ser executado.

```python
for i in range(0,5):
  print(i)

print(list(range(0, 5)))
```

```python
for letter in "Python":
  print(letter)
```

```python
a = 2
resultado = 0

for i in range(10):
  resultado += a
  print("2 x",i,"=", resultado)

i += 1
print("2 x",i,"=", resultado)
```

```python
amigos = ["Joey", "Monica", "Chandler", "Rachel", "Phoebe", "Ross"]

for amigo in amigos:
  print("Olá " + amigo + "! Por favor vem à minha festa!")
```

###**Enumerate**
* No nosso ciclo for podemos muitas vezes querer saber qual é o índice ou posição em que estamos atualmente
* Ao envés de criar uma variável e incrementá-la dentro do ciclo podes usar o **enumerate()**

```python
for posicao, letter in enumerate("Python"):
  print("A letra na posição", posicao, "é:", letter)
```

###**Break**
* Podemos ter certas condições para parar o ciclo quando quisermos.
* A instrução `break` faz parar o ciclo em que estivermos.
* Funciona em ciclos `for` e `while`, bem como em funções e outras estruturas (próximos workshops).



```python
num = 0

while (True):
  if num == 10:
    break
  num += 1
  print(num)
```

###**Continue**
* Permite saltar uma iteração do ciclo.
* O ciclo continua a correr da mesma forma, mas não é corrido o código a seguir a esta instrução.


```python
amigos = ["Joey", "Monica", "Chandler", "Rachel", "Phoebe", "Ross"]

for amigo in amigos:
  print("---Nova iteração---")
  if amigo == "Rachel":
    continue
  print("Olá " + amigo)
```
