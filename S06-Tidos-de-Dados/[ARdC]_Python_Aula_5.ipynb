# **Introdução**

**O que são tipos de dados compostos?**<br>
* São tipos de dados **compostos** em pequenas partes. Por exemplo, as strings são compostas por strings mais pequenas (substrings). No limite, apenas um caractere.

* Tipos de dados simples: `int`, `float`, `bool`, `str`, `complex`

# **Dicionários**

* As listas e os tuplos, ainda que bastante úteis, oferecem uma limitação: apenas permitem acesso sequenciado, ou seja, para acedermos aos seus valores temos sempre de recorrer aos índices onde os respetivos valores se encontram.
* O dicionário surge como solução para esta limitação ao **permitir associar os valores armazenados a uma chave, ou seja, criando um mapeamento entre pares chave/valor**. Tanto as chaves com os valores podem assumir qualquer tipo de dados, existindo apenas a restrição das chaves terem de ser imutáveis.
* Os dicionários não têm qualquer conceito de ordem: os acessos aos valores são sempre feitos pelas chaves, e nunca por índices. 

```python
traduz_pt_en = {
    "um": "one",
    "dois": "two"
}

traduz_pt_en["três"] = "three"

print(traduz_pt_en)
print("Dois em inglês: " + traduz_pt_en["dois"])
```

```python
traduz_pt_en[4] = "four"
print(traduz_pt_en)
```

```python
traduz_pt_en[list()] = "invalid"
```

**Resumindo**.... Um dicionário tem:

- keys()
- values()
- items()

```python
apple_prices = {'macbook': 1000, 'iphone': 500}

```

### **Operações em Dicionários**
* Podemos eliminar um par chave/valor de um dicionário com a instrução **del**.

```python
del traduz_pt_en["dois"]
print(traduz_pt_en)
```

* Podemos a qualquer momento alterar o valor associado a uma chave.

```python
traduz_pt_en["três"] = "nine"
print(traduz_pt_en)
```

* Podemos saber o número de pares chave/valor que um dicionário tem através da função **len**.

```python
len(traduz_pt_en)
```

* Podemos ainda verificar se uma chave existe ou não num dicionário através dos operadores **in** e **not in**, respetivamente.

```python
print("três" in traduz_pt_en)
print("sete" in traduz_pt_en)
print("um" not in traduz_pt_en)
```

### **Métodos dos Dicionários**
* Existe um conjunto de funções associadas ao dicionário, que chamamos de métodos, e que nos oferecem funcionalidades adicionais para esta estrutura de dados.
* É comum querermos iterar sobre todos os pares chave/valor que um dicionário tem. Uma forma simples de o fazermos é iterar pelas chaves através de um ciclo *for*. Para tal, podemos obter a listagem das chaves utilizando o método **keys()**, mas também podemos omitir a chamada a este método.

```python
for chave in traduz_pt_en.keys():
  print("A chave " + str(chave) + " está associada ao valor " + traduz_pt_en[chave])
```

```python
for chave in traduz_pt_en:
  print("A chave " + str(chave) + " está associada ao valor " + traduz_pt_en[chave])
```

* Podemos também obter uma listagem de todos os valores armazenados no dicionário através do método **values()**, ou uma listagem dos pares chave/valor em forma de tuplos através do método **items()**.

```python
traduz_pt_en.values()
```

```python
traduz_pt_en.items()
```

```python
for chave, valor in traduz_pt_en.items():
  print("A chave " + str(chave) + " está associada ao valor " + valor)
```

# **Conjuntos - set**

* Um conjunto é uma estrutura de dados **sem ordem** capaz de armazenar **valores não repetidos**.
* Podemos utilizar o método **.add()** para adicionar elementos a um conjunto e o método **.remove()** para os remover.

```python
numeros_unicos = set([1, 4, 2, 3, 4])
print(numeros_unicos)
```

```python
numeros_unicos.add(7)
numeros_unicos.add(9)
numeros_unicos.add(7)
numeros_unicos.remove(2)
print(numeros_unicos)
```

* Podemos iterar pelos elementos de um conjunto através de um ciclo *for*, e verificar se este contem ou não valores específicos através dos operadores **in** e **not in**.

```python
for valor in numeros_unicos:
  print("O conjunto contem o valor " + str(valor))
```

```python
print(4 in numeros_unicos)
print(20 in numeros_unicos)
print(1 not in numeros_unicos)
```

* Tal como na matemática, podemos aplicar diversas operações aos conjuntos, tais como a reunião e a interseção.

```python
conj1 = set([1, 2, 3, 4])
conj2 = set([3, 4, 5, 6])
print(conj1 | conj2) # Reunião
print(conj1 & conj2) # Interseção
print(conj1 - conj2) # Subtração
```

* Os conjuntos são mutáveis. No entanto, existe uma alternativa imutável (**frozenset**) que não permite adicionar nem remover elementos.

```python
numeros_unicos = frozenset([1, 4, 2, 3, 4])
print(numeros_unicos)
numeros_unicos.add(10)
```

##**Trabalhar com substrings (partes de strings):** Indexing e Slicing
As strings guardam uma **sequência** de caracteres.

Assim, é possível aceder a caracteres individuais (*indexing*) ou subsequências de caracteres de uma string (*slicing*).

*   **Indexing:** Seleciona um único caractere de uma string. Nesse caso é necessário especificar o **índice** que se quer obter! Os índices começam de 0 e especificam a posição da string que se pretende retornar. O que é devolvido é também uma string (no Python não existe `char`).

*   **Slicing:** Seleciona um conjunto (uma *slice*) da string. Nesse caso é necessário especificar o início e o fim do que se pretende selecionar. Neste caso teremos o conjunto **`[n : m]`, que inclui `n` mas exlui `m`**




```python
# "b  a  n  a  n  a"
#  0  1  2  3  4  5 
# -6 -5 -4 -3 -2 -1 

fruta = "banana"
fruta[3]
#fruta[-6]

```

```python
fruta[0:3]
```

### **Comparações de Strings**


*   O operador `==` funciona como operador de igualdade para strings.
*   Os operadores `<` e `>` funcionam para verificar a ordem alfabética!
  * **Nota:** As maiúsculas vêm antes das minúsculas!




```python
palavra = "Swift"

if palavra == 'Python':
  print("As palavras são iguais")
elif palavra > 'Python':
  print("A palavra vem depois de 'Python'")
else:
  print("A palavra vem antes de 'Python'")
```
